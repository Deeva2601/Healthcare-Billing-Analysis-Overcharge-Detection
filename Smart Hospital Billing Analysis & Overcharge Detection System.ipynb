This project focuses on healthcare billing analytics and identifies potential overcharged patients using billing deviation analysis.

The dashboard provides insights into:
- Patient billing distribution
- Hospital revenue trends
- Medical condition billing patterns
- Admission type analysis
- Potential overcharged cases

## Tools & Technologies

- Python
- Pandas
- Matplotlib
- Google Colab
- Power BI
- Excel

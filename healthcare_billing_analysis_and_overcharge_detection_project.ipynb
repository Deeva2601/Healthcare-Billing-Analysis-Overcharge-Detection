**Average Cost per Medical Condition**

```python
df.groupby('Medical Condition')['Billing Amount'].mean().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
avg_cost = df.groupby('Medical Condition')['Billing Amount'].mean().sort_values(ascending=False)
avg_cost.plot(kind='bar',figsize=(10,5))
plt.title('Average Cost per Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Average Billing')
plt.show()
```

```python
df['Difference'] = df['Billing Amount'] - df['Medical Condition'].map(avg_cost)
df['Difference']
```

```python
disease_stats = df.groupby('Medical Condition')['Billing Amount'].agg(['min','max','mean'])
disease_stats
```

```python
df = df.drop(columns=['Expected_min','Expected_max','Expected_Avg','Difference','Overcharge_Flag'], errors='ignore')
```

```python
disease_stats = df.groupby('Medical Condition')['Billing Amount'].agg(
    Expected_min='min',
    Expected_max='max',
    Expected_Avg='mean'
).reset_index()

df = df.merge(disease_stats, on='Medical Condition', how='left')

df['Difference'] = df['Billing Amount'] - df['Expected_Avg']

df['Overcharge_Flag'] = 'Normal'
df.loc[df['Billing Amount'] > df['Expected_Avg'], 'Overcharge_Flag'] = 'Slightly High'
df.loc[df['Billing Amount'] > df['Expected_Avg'] * 1.3, 'Overcharge_Flag'] = 'Very High'

df['Overcharge_Flag'].value_counts()
```

```python
ax = df['Overcharge_Flag'].value_counts().plot(kind='bar',color =['green','orange','red'])

for i,v in enumerate(df['Overcharge_Flag'].value_counts()):
  ax.text(i,v,str(v),ha='center',va='bottom')

plt.title('No.of Patient by Overcharge Category')
plt.xlabel('Overcharge Category')
plt.ylabel('No.of Patient')

plt.show()
```

```python
counts = df['Overcharge_Flag'].value_counts().reindex(['Normal','Slightly High','Very High'])
print(counts)
```
